```python
customers = pd.read_csv('/customers_raw (1).csv')
products = pd.read_csv('/products_raw (1).csv')
orders = pd.read_csv('/orders_raw (1).csv')
order_items = pd.read_csv('/order_items_raw (1).csv')
```

```python
import numpy as np

# ---- CUSTOMERS ----
customers = customers.drop_duplicates()
customers['signup_date'] = pd.to_datetime(customers['signup_date'], format='mixed')
customers['customer_id'] = customers['customer_id'].str.upper()
customers['city'] = customers['city'].str.strip().str.title()
customers['customer_name'] = customers['customer_name'].str.strip()
customers['email'] = customers['email'].replace('', np.nan)
customers['has_email'] = customers['email'].notna()
customers['has_phone'] = customers['phone'].notna()

# ---- PRODUCTS ----
products = products.drop_duplicates()
products['category'] = products['category'].str.strip().str.title()
products.loc[products['unit_price'] < 0, 'unit_price'] = products.loc[products['unit_price'] < 0, 'unit_price'].abs()
median_by_cat = products.groupby('category')['unit_price'].transform('median')
outlier_mask = products['unit_price'] > (median_by_cat * 5)
products.loc[outlier_mask, 'unit_price'] = products.loc[outlier_mask, 'unit_price'] / 10
products['price_was_imputed'] = products['unit_price'].isna()
products['unit_price'] = products['unit_price'].fillna(products.groupby('category')['unit_price'].transform('median'))

# ---- ORDERS ----
orders = orders.drop_duplicates()
orders['order_date'] = pd.to_datetime(orders['order_date'], format='mixed')
orders['customer_id'] = orders['customer_id'].str.upper()   # fix casing BEFORE the FK check
orders['payment_method'] = orders['payment_method'].fillna('Unknown')
orders = orders[orders['customer_id'].isin(customers['customer_id'])]

print("customers:", customers.shape)
print("products:", products.shape)
print("orders:", orders.shape)
```

```python
order_items = order_items.drop_duplicates()
order_items['quantity_was_imputed'] = order_items['quantity'].isna()
order_items['quantity'] = order_items['quantity'].fillna(order_items['quantity'].median())
order_items = order_items[order_items['product_id'].isin(products['product_id'])]  # drop the orphan PR999 row

print("order_items:", order_items.shape)
```
